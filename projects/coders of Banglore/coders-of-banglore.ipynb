```python
chunks = data.split("\n\n")
```

```python
def parce_chuck(chunk):
    ch = chunk.strip()
    split_data = ch.split("\n")
    if(len(split_data) >=4):
        
        username = split_data[0]
        no_of_posts = split_data[1]
        no_of_followers = split_data[2]
        no_of_following = split_data[3]
        name = split_data[4]
        if(len(split_data)>5):
            type_of_page = split_data[5]
            bio = "\n".join(split_data[6:])
        else:
            type_of_page = "Unknown"
            bio = ""
            
        return {
            "username" : username,
            "no_of_posts" : no_of_posts,
            "no_of_followers" : no_of_followers,
            "no_of_following" : no_of_following,
            "name" : name,
            "type_of_page" : type_of_page,
            "bio" : bio
        }
    return {
        "username": "",
        "no_of_posts": "",
        "no_of_followers": "",
        "no_of_following": "",
        "name": "",
        "type_of_page": "Unknown",
        "bio": ""
    }
```

```python
all_chunks = []
for chunk in chunks:
    all_chunks.append(parce_chuck(chunk))
all_chunks
```
