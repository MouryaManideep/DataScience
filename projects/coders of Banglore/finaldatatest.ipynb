```python
import json
```

```python
with open("finaldata.txt",'r',encoding='utf-8') as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if(len(c)>3)]
```

```python
chunks[5]
```

```python
def parce_chuck(chunk):
    ch = chunk.strip()
    split_data = ch.split("\n")
    if(len(split_data) >=4):
        
        username = split_data[0]
        
        no_of_posts = int(split_data[1].split(" ")[0].replace(",",""))
        
        no_of_followers = float(split_data[2].split(" ")[0].replace(",","").replace("K","").replace("M",""))
        if("K" in split_data[2]):
            no_of_followers = int(no_of_followers*1000)
        elif("M" in split_data[2]):
            no_of_followers = int(no_of_followers*1000000)
        else:
            no_of_followers = int(no_of_followers)
            
        no_of_following = float(split_data[3].split(" ")[0].replace(",","").replace("K","").replace("M",""))
        if("K" in split_data[3]):
            no_of_following = int(no_of_following*1000)
        elif("M" in split_data[3]):
            no_of_following = int(no_of_following*1000000)
        else:
            no_of_following = int(no_of_following)
        
        name = split_data[4]
        if(len(split_data)>5):
            type_of_page = split_data[5]
            bio = "\n".join(split_data[6:])
        else:
            type_of_page = "Unknown"
            bio = ""
            
        return {
            "username" : username,
            "no_of_posts" : no_of_posts,
            "no_of_followers" : no_of_followers,
            "no_of_following" : no_of_following,
            "name" : name,
            "type_of_page" : type_of_page,
            "bio" : bio
        }
    return {
        "username": "",
        "no_of_posts": "",
        "no_of_followers": "",
        "no_of_following": "",
        "name": "",
        "type_of_page": "Unknown",
        "bio": ""
    }
```

```python
all_chunks = []
for chunk in chunks:
    all_chunks.append(parce_chuck(chunk))
all_chunks
```

```python
import json
s = json.dumps(all_chunks, indent=4)
with open("finaldata.json", "w") as f:
    f.write(s)
print(s)
```

#### Who has Maximum posts?

```python
max_post = 0
for chunk in all_chunks:
    if(max_post<chunk['no_of_posts']):
        max_post = chunk['no_of_posts']
        max_post_chunk = chunk
print(max_post_chunk)
```

#### Who has Maximum followers?

```python
max_followers = 0
for chunk in all_chunks:
    if(max_followers<chunk['no_of_followers']):
        max_followers = chunk['no_of_followers']
        max_followers_chunk = chunk
print(max_followers_chunk)
```

#### Who has Maximum following?

```python
max_following = 0
for chunk in all_chunks:
    if(max_following < chunk['no_of_following']):
        max_following = chunk['no_of_following']
        max_following_chunk = chunk
print(max_following_chunk)
```
